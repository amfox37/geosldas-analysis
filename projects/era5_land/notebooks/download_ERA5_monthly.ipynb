```python
# era5_monthly_download.py
import cdsapi, time, os, xarray as xr
```

```python

DATASET = "reanalysis-era5-single-levels-monthly-means"

# These variable strings should work the same way in CDS for ERA5 monthly means.
VARS = [
    "soil_temperature_level_1",
    "snow_depth",
    "snow_density",
    "volumetric_soil_water_layer_1",
    "volumetric_soil_water_layer_2",
    "volumetric_soil_water_layer_3",
]
YEARS = list(range(2000, 2024))  # edit as needed
MONTHS = [f"{m:02d}" for m in range(1, 13)]

# Optional: subset to a region (N, W, S, E). Comment out 'area' to get global.
AREA = None  # e.g., [60, -130, 20, -60]

# (optional) rename output dir so you don't mix products
OUT_DIR = "/Users/amfox/Desktop/geosldas-analysis/projects/era5_land/data/era5_monthly_nc"
os.makedirs(OUT_DIR, exist_ok=True)
```

```python

def retrieve_year(year, attempt=1, max_attempts=5):
    c = cdsapi.Client(timeout=600)
    req = {
        "product_type": "monthly_averaged_reanalysis",
        "variable": VARS,
        "year": f"{year}",
        "month": MONTHS,
        "time": "00:00",      # monthly means carry a nominal time
        "data_format": "netcdf",   # CDS supports NetCDF for many ERA5 requests
        # Optional: force grid (ERA5 monthly means are commonly delivered on 0.25°)
        # "grid": [0.25, 0.25],
    }
    if AREA:
        req["area"] = AREA    # N, W, S, E (CDS order)

    target = os.path.join(OUT_DIR, f"era5_monthly_{year}.nc")
    if os.path.exists(target):
        print(f"[skip] {target} exists.")
        return target

    try:
        print(f"[CDS] Requesting {year} ...")
        c.retrieve(DATASET, req, target)
        print(f"[ok] Wrote {target}")
        return target
    except Exception as e:
        if attempt < max_attempts:
            wait = 30 * attempt
            print(f"[warn] {e}\nRetrying year {year} in {wait}s (attempt {attempt+1}/{max_attempts})")
            time.sleep(wait)
            return retrieve_year(year, attempt+1, max_attempts)
        raise
```

```python

# 1) retrieve year-by-year
paths = [retrieve_year(y) for y in YEARS]
```

# Optional alternative merge: `cdo -O mergetime era5_monthly_*.nc era5_monthly_merged.nc`


```python

# 2) merge to a single file
print("[merge] Concatenating yearly NetCDFs ...")
dsets = [xr.open_dataset(p) for p in paths]
ds = xr.concat(dsets, dim="time")
for d in dsets:
    d.close()

# 3) write one consolidated file
out_all = os.path.join(OUT_DIR, f"era5_monthly_{YEARS[0]}_{YEARS[-1]}.nc")
encoding = {v: {"zlib": True, "complevel": 4} for v in ds.data_vars}
encoding.update({c: {"zlib": True, "complevel": 4} for c in ["latitude", "longitude"] if c in ds.coords})
ds.to_netcdf(out_all, encoding=encoding)
print(f"[done] {out_all}")
```

```python
# 4) quick sanity check
print("Variables:", list(ds.data_vars))
print("Time span:", str(ds['time'].values[0])[:10], "to", str(ds['time'].values[-1])[:10])
print("Grid:", ds.sizes.get('latitude', None), "x", ds.sizes.get('longitude', None))
```
