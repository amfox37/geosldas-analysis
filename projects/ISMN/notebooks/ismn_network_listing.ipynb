```python
# Match ISMN USCRN stations to existing USCRN (.mat) sites by nearest lat/lon
import numpy as np
import pandas as pd
import scipy.io as sio
from pathlib import Path

mat_path = Path('/Users/amfox/Desktop/GEOSldas_diagnostics/test_data/M21C_land_sweeper_v2/LS_OLv8_M36_USCRN_SM_1d_c1234smv_25yr_raw_timeseries.mat')
if not mat_path.exists():
    raise FileNotFoundError(f'MAT file not found: {mat_path}')

mat = sio.loadmat(mat_path, squeeze_me=True, struct_as_record=False)


def _to_1d_array(x, n_site=None):
    arr = np.asarray(x)
    if arr.dtype == object and arr.size == 1:
        arr = np.asarray(arr.item())
    arr = np.squeeze(arr)
    if arr.ndim == 0:
        arr = np.array([arr])
    if arr.ndim > 1:
        if n_site is not None and n_site in arr.shape:
            axis = int(np.where(np.array(arr.shape) == n_site)[0][0])
            arr = np.moveaxis(arr, axis, 0).reshape(n_site, -1)
            if arr.shape[1] == 1:
                arr = arr[:, 0]
        else:
            arr = arr.ravel()
    return np.asarray(arr)


def _normalize_site_ids(raw_ids, n_site):
    arr = _to_1d_array(raw_ids, n_site=n_site)
    if arr.size != n_site:
        arr = np.resize(arr, n_site)

    out = []
    for x in arr:
        if isinstance(x, bytes):
            s = x.decode('utf-8', errors='ignore').strip()
        elif isinstance(x, str):
            s = x.strip()
        else:
            try:
                v = float(x)
                if np.isfinite(v) and abs(v - round(v)) < 1e-8:
                    s = str(int(round(v)))
                else:
                    s = str(v)
            except Exception:
                s = str(x)
        out.append(s)
    return np.array(out, dtype=object)


def _extract_existing_site_meta(mat_dict):
    n_site = None
    if 'insitu' in mat_dict:
        arr = np.asarray(mat_dict['insitu'])
        if arr.ndim >= 3:
            n_site = arr.shape[2]
    if n_site is None:
        for key in ['INSITU_lat', 'site_lat', 'lat', 'LAT']:
            if key in mat_dict:
                n_site = _to_1d_array(mat_dict[key]).size
                break
    if n_site is None:
        raise RuntimeError('Could not infer n_site from MAT file')

    lat = np.full(n_site, np.nan, dtype=float)
    lon = np.full(n_site, np.nan, dtype=float)
    ids = np.array([str(i) for i in range(n_site)], dtype=object)

    for key in ['INSITU_lat', 'site_lat', 'lat', 'LAT']:
        if key in mat_dict:
            try:
                tmp = _to_1d_array(mat_dict[key], n_site=n_site).astype(float)
                if tmp.size == n_site:
                    lat = tmp
                    break
            except Exception:
                pass

    for key in ['INSITU_lon', 'site_lon', 'lon', 'LON']:
        if key in mat_dict:
            try:
                tmp = _to_1d_array(mat_dict[key], n_site=n_site).astype(float)
                if tmp.size == n_site:
                    lon = tmp
                    break
            except Exception:
                pass

    for key in ['INSITU_id', 'site_id', 'id', 'ID']:
        if key in mat_dict:
            try:
                ids = _normalize_site_ids(mat_dict[key], n_site)
                break
            except Exception:
                pass

    return pd.DataFrame({
        'existing_site_index': np.arange(n_site, dtype=int),
        'existing_site_id': ids,
        'existing_lat': lat,
        'existing_lon': lon,
    })


def _get_meta_series(df, field_name):
    if df is None:
        return None
    target = field_name.lower()

    # MultiIndex columns: (field, attr), e.g. ('station','val')
    if isinstance(df.columns, pd.MultiIndex):
        lvl0 = np.array([str(x).lower() for x in df.columns.get_level_values(0)])
        cols = df.columns[lvl0 == target]
        if len(cols) == 0:
            return None
        sub = df.loc[:, cols]
        if isinstance(sub, pd.Series):
            return sub
        if isinstance(sub.columns, pd.MultiIndex):
            lvl1 = [str(x).lower() for x in sub.columns.get_level_values(1)]
            for key in ['val', 'value', 'name']:
                if key in lvl1:
                    return sub.iloc[:, lvl1.index(key)]
        return sub.iloc[:, 0]

    # Flat columns
    lower_cols = {str(c).lower(): c for c in df.columns}
    if target in lower_cols:
        return df[lower_cols[target]]

    # Tuple-like flat columns
    for c in df.columns:
        cs = str(c).lower()
        if cs.startswith(f"('{target}',") or cs.startswith(f'("{target}",'):
            return df[c]

    return None


def _extract_station_lat_lon_from_sensor(st):
    # Fallback path when station-level metadata is sparse
    if not hasattr(st, 'iter_sensors'):
        return np.nan, np.nan

    sensors = []
    try:
        sensors = list(st.iter_sensors())
    except Exception:
        sensors = []

    for sensor in sensors:
        # direct attrs
        for lat_attr, lon_attr in [('latitude', 'longitude'), ('lat', 'lon')]:
            try:
                lat = float(getattr(sensor, lat_attr))
                lon = float(getattr(sensor, lon_attr))
                if np.isfinite(lat) and np.isfinite(lon):
                    return lat, lon
            except Exception:
                pass

        # sensor metadata
        md = getattr(sensor, 'metadata', None)
        mdf = None
        if md is not None:
            try:
                mdf = md.to_pd() if hasattr(md, 'to_pd') else md
            except Exception:
                mdf = None
        if isinstance(mdf, pd.DataFrame):
            lat_s = _get_meta_series(mdf, 'latitude')
            lon_s = _get_meta_series(mdf, 'longitude')
            if lat_s is None:
                lat_s = _get_meta_series(mdf, 'lat')
            if lon_s is None:
                lon_s = _get_meta_series(mdf, 'lon')
            if lat_s is not None and lon_s is not None:
                lat_v = pd.to_numeric(pd.Series(lat_s), errors='coerce').dropna()
                lon_v = pd.to_numeric(pd.Series(lon_s), errors='coerce').dropna()
                if len(lat_v) > 0 and len(lon_v) > 0:
                    return float(lat_v.iloc[0]), float(lon_v.iloc[0])

    return np.nan, np.nan


def _extract_ismn_uscrn_station_meta(ismn_obj):
    # Preferred: collection metadata table
    mdf = getattr(ismn_obj, 'metadata', None)
    if mdf is not None and not isinstance(mdf, pd.DataFrame):
        try:
            mdf = mdf.to_pd() if hasattr(mdf, 'to_pd') else mdf
        except Exception:
            mdf = None

    if isinstance(mdf, pd.DataFrame):
        station_s = _get_meta_series(mdf, 'station')
        network_s = _get_meta_series(mdf, 'network')
        lat_s = _get_meta_series(mdf, 'latitude')
        lon_s = _get_meta_series(mdf, 'longitude')

        if lat_s is None:
            lat_s = _get_meta_series(mdf, 'lat')
        if lon_s is None:
            lon_s = _get_meta_series(mdf, 'lon')

        if station_s is not None and network_s is not None and lat_s is not None and lon_s is not None:
            meta_df = pd.DataFrame({
                'ismn_station_name': pd.Series(station_s).astype(str),
                'ismn_network': pd.Series(network_s).astype(str),
                'ismn_lat': pd.to_numeric(pd.Series(lat_s), errors='coerce'),
                'ismn_lon': pd.to_numeric(pd.Series(lon_s), errors='coerce'),
            })
            meta_df = meta_df[meta_df['ismn_network'].str.upper() == 'USCRN'].copy()
            if len(meta_df) > 0:
                # one row per station
                grouped = (
                    meta_df
                    .groupby('ismn_station_name', as_index=False)
                    .agg({'ismn_lat': 'median', 'ismn_lon': 'median'})
                    .sort_values('ismn_station_name')
                    .reset_index(drop=True)
                )
                return grouped

    # Fallback: station objects + sensor metadata
    if 'USCRN' not in ismn_obj.networks:
        raise KeyError('USCRN network not found in ISMN data')

    uscrn_net = ismn_obj['USCRN']
    rows = []
    for station_name in uscrn_net.stations:
        st = uscrn_net[station_name]

        lat = np.nan
        lon = np.nan
        # direct station attrs
        for lat_attr, lon_attr in [('latitude', 'longitude'), ('lat', 'lon')]:
            try:
                lat = float(getattr(st, lat_attr))
                lon = float(getattr(st, lon_attr))
                if np.isfinite(lat) and np.isfinite(lon):
                    break
            except Exception:
                pass

        if not (np.isfinite(lat) and np.isfinite(lon)):
            lat, lon = _extract_station_lat_lon_from_sensor(st)

        rows.append({'ismn_station_name': station_name, 'ismn_lat': lat, 'ismn_lon': lon})

    return pd.DataFrame(rows).sort_values('ismn_station_name').reset_index(drop=True)


def _haversine_km(lat1, lon1, lat2, lon2):
    r = 6371.0
    p1 = np.deg2rad(lat1)
    p2 = np.deg2rad(lat2)
    dphi = np.deg2rad(lat2 - lat1)
    dlambda = np.deg2rad(lon2 - lon1)
    a = np.sin(dphi / 2.0) ** 2 + np.cos(p1) * np.cos(p2) * np.sin(dlambda / 2.0) ** 2
    return 2.0 * r * np.arcsin(np.sqrt(a))


ismn_df = _extract_ismn_uscrn_station_meta(ismn_data)
existing_df = _extract_existing_site_meta(mat)

finite_ismn = np.isfinite(ismn_df['ismn_lat'].to_numpy()) & np.isfinite(ismn_df['ismn_lon'].to_numpy())
finite_existing = np.isfinite(existing_df['existing_lat'].to_numpy()) & np.isfinite(existing_df['existing_lon'].to_numpy())

print(f'ISMN USCRN stations: {len(ismn_df)} (finite lat/lon: {int(finite_ismn.sum())})')
print(f'Existing USCRN sites from MAT: {len(existing_df)} (finite lat/lon: {int(finite_existing.sum())})')

if not finite_existing.any():
    raise RuntimeError('No finite lat/lon found in existing MAT site metadata')

ex = existing_df.loc[finite_existing].reset_index(drop=True)
ex_lat = ex['existing_lat'].to_numpy(dtype=float)
ex_lon = ex['existing_lon'].to_numpy(dtype=float)

matched_rows = []
for _, row in ismn_df.iterrows():
    lat = row['ismn_lat']
    lon = row['ismn_lon']
    if not (np.isfinite(lat) and np.isfinite(lon)):
        matched_rows.append({
            **row.to_dict(),
            'existing_site_index': np.nan,
            'existing_site_id': np.nan,
            'existing_lat': np.nan,
            'existing_lon': np.nan,
            'match_distance_km': np.nan,
        })
        continue

    d = _haversine_km(lat, lon, ex_lat, ex_lon)
    j = int(np.argmin(d))
    matched_rows.append({
        **row.to_dict(),
        'existing_site_index': int(ex.iloc[j]['existing_site_index']),
        'existing_site_id': ex.iloc[j]['existing_site_id'],
        'existing_lat': float(ex.iloc[j]['existing_lat']),
        'existing_lon': float(ex.iloc[j]['existing_lon']),
        'match_distance_km': float(d[j]),
    })

MATCH_MAX_KM = 2.0

raw_match_df = (
    pd.DataFrame(matched_rows)
    .sort_values(['match_distance_km', 'ismn_station_name'], na_position='last')
    .reset_index(drop=True)
)

match_df = (
    raw_match_df[
        np.isfinite(raw_match_df['match_distance_km'])
        & (raw_match_df['match_distance_km'] <= MATCH_MAX_KM)
    ]
    .copy()
    .reset_index(drop=True)
)

print(f'Matching threshold: <= {MATCH_MAX_KM:.1f} km')
print(f'USCRN matches kept: {len(match_df)} / {len(raw_match_df)}')
print('Table columns: ISMN station + lat/lon, matched existing site id + lat/lon, distance (km)')
display(match_df)


```

```python
# Random matched-site comparison: ISMN vs MAT Obs vs OL (.mat) surface soil moisture (native timesteps)
# + focused 2020-2021 view: ISMN native vs ISMN daily vs MAT Obs native
import re
import datetime as dt
import numpy as np
import pandas as pd
import scipy.io as sio
import matplotlib.pyplot as plt

if 'match_df' not in globals():
    raise RuntimeError('Run the matching cell first (match_df is missing).')

# -------------------------
# user knobs
# -------------------------
N_RANDOM_SITES = 5
RANDOM_SEED = 42
MAX_MATCH_KM = 2.0
TARGET_SURFACE_DEPTH_M = 0.05
OL_MAT_PATH = Path('/Users/amfox/Desktop/GEOSldas_diagnostics/test_data/M21C_land_sweeper_v2/LS_OLv8_M36_USCRN_SM_1d_c1234smv_25yr_raw_timeseries.mat')
ZOOM_START = pd.Timestamp('2021-01-01')
ZOOM_END = pd.Timestamp('2022-12-31 23:59:59')

if not OL_MAT_PATH.exists():
    raise FileNotFoundError(f'OL MAT file not found: {OL_MAT_PATH}')


def _mat_struct_to_array(ms, field):
    if isinstance(ms, np.ndarray):
        flat = ms.ravel()
        return np.array([getattr(x, field) for x in flat])
    return np.array([getattr(ms, field)])


def _build_timestamps(dtv):
    year = _mat_struct_to_array(dtv, 'year').astype(int).ravel()
    month = _mat_struct_to_array(dtv, 'month').astype(int).ravel()
    day = _mat_struct_to_array(dtv, 'day').astype(int).ravel()
    try:
        hour = _mat_struct_to_array(dtv, 'hour').astype(int).ravel()
    except Exception:
        hour = np.zeros_like(year)
    try:
        minute = _mat_struct_to_array(dtv, 'min').astype(int).ravel()
    except Exception:
        minute = np.zeros_like(year)
    try:
        second = _mat_struct_to_array(dtv, 'sec').astype(int).ravel()
    except Exception:
        second = np.zeros_like(year)

    return pd.to_datetime([
        dt.datetime(int(y), int(m), int(d), int(h), int(mi), int(s))
        for y, m, d, h, mi, s in zip(year, month, day, hour, minute, second)
    ])


def _extract_depth_from_name(sensor_name):
    # typical pattern: ..._soil_moisture_0.050000_0.050000
    m = re.search(r'_(-?\d+\.\d+)_(-?\d+\.\d+)$', str(sensor_name))
    if not m:
        return np.nan
    d1 = float(m.group(1))
    d2 = float(m.group(2))
    return 0.5 * (d1 + d2)


def _pick_surface_sm_sensor(station_obj, target_depth=0.05):
    sensors = []
    if hasattr(station_obj, 'iter_sensors'):
        try:
            sensors = list(station_obj.iter_sensors(variable='soil_moisture'))
        except TypeError:
            sensors = [s for s in station_obj.iter_sensors() if 'soil_moisture' in getattr(s, 'name', '').lower()]

    if len(sensors) == 0:
        raise RuntimeError(f'No soil moisture sensors found at station {station_obj}')

    depths = np.array([_extract_depth_from_name(getattr(s, 'name', '')) for s in sensors], dtype=float)
    if np.any(np.isfinite(depths)):
        j = int(np.nanargmin(np.abs(depths - float(target_depth))))
    else:
        j = 0
    return sensors[j]


def _sensor_dataframe(sensor_obj):
    try:
        df = sensor_obj.data
    except Exception:
        df = sensor_obj.read_data()

    if not isinstance(df, pd.DataFrame):
        df = pd.DataFrame(df)

    if not isinstance(df.index, pd.DatetimeIndex):
        try:
            df.index = pd.to_datetime(df.index)
        except Exception:
            raise RuntimeError('Sensor data index is not datetime-like and could not be converted')
    return df


def _pick_var_col(df, var_name):
    cols = [c for c in df.columns if var_name in str(c).lower()]
    if len(cols) > 0:
        return cols[0]
    num_cols = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]
    if len(num_cols) == 0:
        return None
    return num_cols[0]


def _sensor_series_qc(sm_sensor_obj, required_sm_flag='G'):
    df_sm = _sensor_dataframe(sm_sensor_obj)

    sm_col = _pick_var_col(df_sm, 'soil_moisture')
    if sm_col is None:
        raise RuntimeError(f'No soil moisture numeric column found for sensor {getattr(sm_sensor_obj, "name", "")}')

    flag_cols = [c for c in df_sm.columns if str(c).lower() == 'soil_moisture_flag']
    if len(flag_cols) == 0:
        flag_cols = [c for c in df_sm.columns if 'soil_moisture_flag' in str(c).lower()]
    if len(flag_cols) == 0:
        raise RuntimeError(f'No soil_moisture_flag column found for sensor {getattr(sm_sensor_obj, "name", "")}')

    sm = pd.to_numeric(df_sm[sm_col], errors='coerce').copy()
    flag = df_sm[flag_cols[0]].astype(str).str.strip().str.upper()
    keep_mask = flag.eq(str(required_sm_flag).strip().upper())

    s = sm.where(keep_mask)
    s = s.groupby(s.index).mean().sort_index()
    return s


# load MAT data once
mat_ol = sio.loadmat(OL_MAT_PATH, squeeze_me=True, struct_as_record=False)
required = ['LDAS_sm_org', 'INSITU_sm', 'date_time_vec']
missing = [k for k in required if k not in mat_ol]
if missing:
    raise KeyError(f'OL MAT missing required keys: {missing}')

ldas = np.array(mat_ol['LDAS_sm_org'], dtype=float)
obs_mat = np.array(mat_ol['INSITU_sm'], dtype=float)

if ldas.ndim == 4:
    ldas = ldas[:, :, :, 0]
if ldas.ndim != 3:
    raise ValueError(f'Unexpected LDAS_sm_org shape: {ldas.shape}')
if obs_mat.ndim != 3:
    raise ValueError(f'Unexpected INSITU_sm shape: {obs_mat.shape}')

# surface depth only
ldas_surface = ldas[:, 0, :].astype(float)
obs_surface_mat = obs_mat[:, 0, :].astype(float)

# common missing-value handling
ldas_surface[ldas_surface == 0] = np.nan
obs_surface_mat[obs_surface_mat < -9000] = np.nan

mat_time = _build_timestamps(mat_ol['date_time_vec'])

# choose random matched stations with usable coordinates
cand = match_df[np.isfinite(match_df['match_distance_km'])].copy()
if 'existing_site_index' not in cand.columns:
    raise RuntimeError('match_df missing existing_site_index column')

close_cand = cand[cand['match_distance_km'] <= MAX_MATCH_KM].copy()
if len(close_cand) >= N_RANDOM_SITES:
    cand = close_cand

if len(cand) == 0:
    raise RuntimeError('No matched stations with finite distance found.')

n_pick = min(N_RANDOM_SITES, len(cand))
rng = np.random.default_rng(RANDOM_SEED)
pick_idx = rng.choice(cand.index.to_numpy(), size=n_pick, replace=False)
sel = cand.loc[pick_idx].copy().sort_values('ismn_station_name').reset_index(drop=True)

print('Selected random stations for ISMN vs MAT Obs vs OL comparison:')
display(sel[[
    'ismn_station_name', 'ismn_lat', 'ismn_lon',
    'existing_site_index', 'existing_site_id', 'existing_lat', 'existing_lon', 'match_distance_km'
]])

# collect series per selected site once (reused across both figures)
site_series = []
for _, row in sel.iterrows():
    station_name = str(row['ismn_station_name'])
    site_idx = int(row['existing_site_index'])

    if site_idx < 0 or site_idx >= ldas_surface.shape[1]:
        raise IndexError(f'existing_site_index {site_idx} out of bounds for MAT surface array')

    st = ismn_data['USCRN'][station_name]
    sm_sensor = _pick_surface_sm_sensor(st, target_depth=TARGET_SURFACE_DEPTH_M)
    s_ismn_native = _sensor_series_qc(sm_sensor, required_sm_flag='G')
    s_ismn_daily = s_ismn_native.resample('1D').mean()
    s_ismn_daily.index = s_ismn_daily.index + pd.Timedelta(hours=12)

    s_ol_native = pd.Series(ldas_surface[:, site_idx], index=mat_time).sort_index()
    s_obs_mat_native = pd.Series(obs_surface_mat[:, site_idx], index=mat_time).sort_index()

    site_series.append({
        'station_name': station_name,
        'site_idx': site_idx,
        'site_id': row['existing_site_id'],
        's_ismn_native': s_ismn_native,
        's_ismn_daily': s_ismn_daily,
        's_obs_mat_native': s_obs_mat_native,
        's_ol_native': s_ol_native,
    })

# -------------------------
# Figure 1: full period (3 lines) -> ISMN native, MAT obs native, OL native
# -------------------------
fig1, axs1 = plt.subplots(n_pick, 1, figsize=(14, 4 * n_pick), sharex=False)
if n_pick == 1:
    axs1 = [axs1]

for ax, dct in zip(axs1, site_series):
    ax.plot(dct['s_ismn_native'].index, dct['s_ismn_native'].values, color='k', lw=0.8, alpha=0.8, label='ISMN Obs (native, G-flag QC)')
    ax.plot(dct['s_obs_mat_native'].index, dct['s_obs_mat_native'].values, color='#4C78A8', lw=1.0, label='MAT Obs (native)')
    ax.plot(dct['s_ol_native'].index, dct['s_ol_native'].values, color='#F28E2B', lw=1.0, label='OL model (native)')

    ax.set_title(f"USCRN {dct['station_name']}  <->  MAT site_id {dct['site_id']} (index {dct['site_idx']})")
    ax.set_ylabel('Surface SM')
    ax.grid(alpha=0.3)
    ax.legend(loc='upper right')

axs1[-1].set_xlabel('Time')
fig1.suptitle('Random matched sites: ISMN Obs vs MAT Obs vs OL (surface SM, native timesteps)', y=1.02)
plt.tight_layout()
plt.show()

# -------------------------
# Figure 2: 2020-2021 zoom (2 lines) -> ISMN daily, MAT obs native
# -------------------------
fig2, axs2 = plt.subplots(n_pick, 1, figsize=(14, 4 * n_pick), sharex=False)
if n_pick == 1:
    axs2 = [axs2]

for ax, dct in zip(axs2, site_series):
    s_ismn_daily_zoom = dct['s_ismn_daily'].loc[(dct['s_ismn_daily'].index >= ZOOM_START) & (dct['s_ismn_daily'].index <= ZOOM_END)]
    s_obs_mat_zoom = dct['s_obs_mat_native'].loc[(dct['s_obs_mat_native'].index >= ZOOM_START) & (dct['s_obs_mat_native'].index <= ZOOM_END)]

    ax.plot(s_ismn_daily_zoom.index, s_ismn_daily_zoom.values, lw=2.5, linestyle=":", label='ISMN Obs (daily mean, G-flag QC)')
    ax.plot(s_obs_mat_zoom.index, s_obs_mat_zoom.values, lw=1.0, label='MAT Obs (native)')

    ax.set_title(f"USCRN {dct['station_name']}  <->  MAT site_id {dct['site_id']} (index {dct['site_idx']}) | 2020-2021")
    ax.set_ylabel('Surface SM')
    ax.grid(alpha=0.3)
    ax.legend(loc='upper right')

axs2[-1].set_xlabel('Time')
fig2.suptitle('Random matched sites: 2020-2021 ISMN daily vs MAT Obs native', y=1.02)
plt.tight_layout()
plt.show()





```

```python
# SCAN: match ISMN stations to existing SCAN (.mat) sites by nearest lat/lon
import numpy as np
import pandas as pd
import scipy.io as sio
from pathlib import Path

SCAN_OL_MAT_PATH = Path('/Users/amfox/Desktop/GEOSldas_diagnostics/test_data/M21C_land_sweeper_v2/LS_OLv8_M36_SCAN_SM_1d_c1234smv_25yr_raw_timeseries.mat')
if not SCAN_OL_MAT_PATH.exists():
    raise FileNotFoundError(f'SCAN MAT file not found: {SCAN_OL_MAT_PATH}')

mat_scan = sio.loadmat(SCAN_OL_MAT_PATH, squeeze_me=True, struct_as_record=False)


def _scan_to_1d_array(x, n_site=None):
    arr = np.asarray(x)
    if arr.dtype == object and arr.size == 1:
        arr = np.asarray(arr.item())
    arr = np.squeeze(arr)
    if arr.ndim == 0:
        arr = np.array([arr])
    if arr.ndim > 1:
        if n_site is not None and n_site in arr.shape:
            axis = int(np.where(np.array(arr.shape) == n_site)[0][0])
            arr = np.moveaxis(arr, axis, 0).reshape(n_site, -1)
            if arr.shape[1] == 1:
                arr = arr[:, 0]
        else:
            arr = arr.ravel()
    return np.asarray(arr)


def _scan_normalize_site_ids(raw_ids, n_site):
    arr = _scan_to_1d_array(raw_ids, n_site=n_site)
    if arr.size != n_site:
        arr = np.resize(arr, n_site)

    out = []
    for x in arr:
        if isinstance(x, bytes):
            s = x.decode('utf-8', errors='ignore').strip()
        elif isinstance(x, str):
            s = x.strip()
        else:
            try:
                v = float(x)
                if np.isfinite(v) and abs(v - round(v)) < 1e-8:
                    s = str(int(round(v)))
                else:
                    s = str(v)
            except Exception:
                s = str(x)
        out.append(s)
    return np.array(out, dtype=object)


def _scan_extract_existing_site_meta(mat_dict):
    n_site = None
    if 'insitu' in mat_dict:
        arr = np.asarray(mat_dict['insitu'])
        if arr.ndim >= 3:
            n_site = arr.shape[2]
    if n_site is None:
        for key in ['INSITU_lat', 'site_lat', 'lat', 'LAT']:
            if key in mat_dict:
                n_site = _scan_to_1d_array(mat_dict[key]).size
                break
    if n_site is None:
        raise RuntimeError('Could not infer n_site from MAT file')

    lat = np.full(n_site, np.nan, dtype=float)
    lon = np.full(n_site, np.nan, dtype=float)
    ids = np.array([str(i) for i in range(n_site)], dtype=object)

    for key in ['INSITU_lat', 'site_lat', 'lat', 'LAT']:
        if key in mat_dict:
            try:
                tmp = _scan_to_1d_array(mat_dict[key], n_site=n_site).astype(float)
                if tmp.size == n_site:
                    lat = tmp
                    break
            except Exception:
                pass

    for key in ['INSITU_lon', 'site_lon', 'lon', 'LON']:
        if key in mat_dict:
            try:
                tmp = _scan_to_1d_array(mat_dict[key], n_site=n_site).astype(float)
                if tmp.size == n_site:
                    lon = tmp
                    break
            except Exception:
                pass

    for key in ['INSITU_id', 'site_id', 'id', 'ID']:
        if key in mat_dict:
            try:
                ids = _scan_normalize_site_ids(mat_dict[key], n_site)
                break
            except Exception:
                pass

    return pd.DataFrame({
        'existing_site_index': np.arange(n_site, dtype=int),
        'existing_site_id': ids,
        'existing_lat': lat,
        'existing_lon': lon,
    })


def _scan_get_meta_series(df, field_name):
    if df is None:
        return None
    target = field_name.lower()

    if isinstance(df.columns, pd.MultiIndex):
        lvl0 = np.array([str(x).lower() for x in df.columns.get_level_values(0)])
        cols = df.columns[lvl0 == target]
        if len(cols) == 0:
            return None
        sub = df.loc[:, cols]
        if isinstance(sub, pd.Series):
            return sub
        if isinstance(sub.columns, pd.MultiIndex):
            lvl1 = [str(x).lower() for x in sub.columns.get_level_values(1)]
            for key in ['val', 'value', 'name']:
                if key in lvl1:
                    return sub.iloc[:, lvl1.index(key)]
        return sub.iloc[:, 0]

    lower_cols = {str(c).lower(): c for c in df.columns}
    if target in lower_cols:
        return df[lower_cols[target]]

    for c in df.columns:
        cs = str(c).lower()
        if cs.startswith(f"('{target}',") or cs.startswith(f'("{target}",'):
            return df[c]

    return None


def _scan_extract_ismn_network_station_meta(ismn_obj, network_name='SCAN'):
    mdf = getattr(ismn_obj, 'metadata', None)
    if mdf is not None and not isinstance(mdf, pd.DataFrame):
        try:
            mdf = mdf.to_pd() if hasattr(mdf, 'to_pd') else mdf
        except Exception:
            mdf = None

    if not isinstance(mdf, pd.DataFrame):
        raise RuntimeError('ismn_data.metadata is not accessible as DataFrame')

    station_s = _scan_get_meta_series(mdf, 'station')
    network_s = _scan_get_meta_series(mdf, 'network')
    lat_s = _scan_get_meta_series(mdf, 'latitude')
    lon_s = _scan_get_meta_series(mdf, 'longitude')

    if lat_s is None:
        lat_s = _scan_get_meta_series(mdf, 'lat')
    if lon_s is None:
        lon_s = _scan_get_meta_series(mdf, 'lon')

    if station_s is None or network_s is None or lat_s is None or lon_s is None:
        raise RuntimeError('Could not find station/network/lat/lon columns in ismn_data.metadata')

    out = pd.DataFrame({
        'ismn_station_name': pd.Series(station_s).astype(str),
        'ismn_network': pd.Series(network_s).astype(str),
        'ismn_lat': pd.to_numeric(pd.Series(lat_s), errors='coerce'),
        'ismn_lon': pd.to_numeric(pd.Series(lon_s), errors='coerce'),
    })

    out = out[out['ismn_network'].str.upper() == str(network_name).upper()].copy()
    out = (
        out.groupby('ismn_station_name', as_index=False)
           .agg({'ismn_lat': 'median', 'ismn_lon': 'median'})
           .sort_values('ismn_station_name')
           .reset_index(drop=True)
    )
    return out


def _scan_haversine_km(lat1, lon1, lat2, lon2):
    r = 6371.0
    p1 = np.deg2rad(lat1)
    p2 = np.deg2rad(lat2)
    dphi = np.deg2rad(lat2 - lat1)
    dlambda = np.deg2rad(lon2 - lon1)
    a = np.sin(dphi / 2.0) ** 2 + np.cos(p1) * np.cos(p2) * np.sin(dlambda / 2.0) ** 2
    return 2.0 * r * np.arcsin(np.sqrt(a))


ismn_scan_df = _scan_extract_ismn_network_station_meta(ismn_data, network_name='SCAN')
existing_scan_df = _scan_extract_existing_site_meta(mat_scan)

finite_ismn_scan = np.isfinite(ismn_scan_df['ismn_lat'].to_numpy()) & np.isfinite(ismn_scan_df['ismn_lon'].to_numpy())
finite_existing_scan = np.isfinite(existing_scan_df['existing_lat'].to_numpy()) & np.isfinite(existing_scan_df['existing_lon'].to_numpy())

print(f'SCAN ISMN stations: {len(ismn_scan_df)} (finite lat/lon: {int(finite_ismn_scan.sum())})')
print(f'Existing SCAN MAT sites: {len(existing_scan_df)} (finite lat/lon: {int(finite_existing_scan.sum())})')

if not finite_existing_scan.any():
    raise RuntimeError('No finite lat/lon found in SCAN MAT site metadata')

ex = existing_scan_df.loc[finite_existing_scan].reset_index(drop=True)
ex_lat = ex['existing_lat'].to_numpy(dtype=float)
ex_lon = ex['existing_lon'].to_numpy(dtype=float)

rows = []
for _, row in ismn_scan_df.iterrows():
    lat = row['ismn_lat']
    lon = row['ismn_lon']
    if not (np.isfinite(lat) and np.isfinite(lon)):
        rows.append({
            **row.to_dict(),
            'existing_site_index': np.nan,
            'existing_site_id': np.nan,
            'existing_lat': np.nan,
            'existing_lon': np.nan,
            'match_distance_km': np.nan,
        })
        continue

    d = _scan_haversine_km(lat, lon, ex_lat, ex_lon)
    j = int(np.argmin(d))
    rows.append({
        **row.to_dict(),
        'existing_site_index': int(ex.iloc[j]['existing_site_index']),
        'existing_site_id': ex.iloc[j]['existing_site_id'],
        'existing_lat': float(ex.iloc[j]['existing_lat']),
        'existing_lon': float(ex.iloc[j]['existing_lon']),
        'match_distance_km': float(d[j]),
    })

SCAN_MATCH_MAX_KM = 2.0

raw_match_df_scan = (
    pd.DataFrame(rows)
    .sort_values(['match_distance_km', 'ismn_station_name'], na_position='last')
    .reset_index(drop=True)
)

match_df_scan = (
    raw_match_df_scan[
        np.isfinite(raw_match_df_scan['match_distance_km'])
        & (raw_match_df_scan['match_distance_km'] <= SCAN_MATCH_MAX_KM)
    ]
    .copy()
    .reset_index(drop=True)
)

print(f'Matching threshold: <= {SCAN_MATCH_MAX_KM:.1f} km')
print(f'SCAN matches kept: {len(match_df_scan)} / {len(raw_match_df_scan)}')
print('Table columns: ISMN SCAN station + lat/lon, matched MAT SCAN site id + lat/lon, distance (km)')
display(match_df_scan)


```

```python
# SCAN random matched-site comparison (same exercise as USCRN)
# Figure 1: full period -> ISMN Obs native, MAT Obs native, OL model native
# Figure 2: 2020-2021 -> ISMN Obs daily (+12h), MAT Obs native
import re
import datetime as dt
import numpy as np
import pandas as pd
import scipy.io as sio
import matplotlib.pyplot as plt

if 'match_df_scan' not in globals():
    raise RuntimeError('Run the SCAN matching cell first (match_df_scan missing).')

# -------------------------
# user knobs
# -------------------------
SCAN_N_RANDOM_SITES = 5
SCAN_RANDOM_SEED = 42
SCAN_MAX_MATCH_KM = 2.0
SCAN_TARGET_SURFACE_DEPTH_M = 0.05
SCAN_OL_MAT_PATH = Path('/Users/amfox/Desktop/GEOSldas_diagnostics/test_data/M21C_land_sweeper_v2/LS_OLv8_M36_SCAN_SM_1d_c1234smv_25yr_raw_timeseries.mat')
SCAN_ZOOM_START = pd.Timestamp('2020-01-01')
SCAN_ZOOM_END = pd.Timestamp('2021-12-31 23:59:59')

if not SCAN_OL_MAT_PATH.exists():
    raise FileNotFoundError(f'SCAN MAT file not found: {SCAN_OL_MAT_PATH}')


def _scan_plot_mat_struct_to_array(ms, field):
    if isinstance(ms, np.ndarray):
        flat = ms.ravel()
        return np.array([getattr(x, field) for x in flat])
    return np.array([getattr(ms, field)])


def _scan_plot_build_timestamps(dtv):
    year = _scan_plot_mat_struct_to_array(dtv, 'year').astype(int).ravel()
    month = _scan_plot_mat_struct_to_array(dtv, 'month').astype(int).ravel()
    day = _scan_plot_mat_struct_to_array(dtv, 'day').astype(int).ravel()
    try:
        hour = _scan_plot_mat_struct_to_array(dtv, 'hour').astype(int).ravel()
    except Exception:
        hour = np.zeros_like(year)
    try:
        minute = _scan_plot_mat_struct_to_array(dtv, 'min').astype(int).ravel()
    except Exception:
        minute = np.zeros_like(year)
    try:
        second = _scan_plot_mat_struct_to_array(dtv, 'sec').astype(int).ravel()
    except Exception:
        second = np.zeros_like(year)

    return pd.to_datetime([
        dt.datetime(int(y), int(m), int(d), int(h), int(mi), int(s))
        for y, m, d, h, mi, s in zip(year, month, day, hour, minute, second)
    ])


def _scan_plot_extract_depth_from_name(sensor_name):
    m = re.search(r'_(-?\d+\.\d+)_(-?\d+\.\d+)$', str(sensor_name))
    if not m:
        return np.nan
    d1 = float(m.group(1))
    d2 = float(m.group(2))
    return 0.5 * (d1 + d2)


def _scan_pick_surface_sm_sensor(station_obj, target_depth=0.05):
    sensors = []
    if hasattr(station_obj, 'iter_sensors'):
        try:
            sensors = list(station_obj.iter_sensors(variable='soil_moisture'))
        except TypeError:
            sensors = [s for s in station_obj.iter_sensors() if 'soil_moisture' in getattr(s, 'name', '').lower()]

    if len(sensors) == 0:
        raise RuntimeError(f'No soil moisture sensors found at station {station_obj}')

    depths = np.array([_scan_plot_extract_depth_from_name(getattr(s, 'name', '')) for s in sensors], dtype=float)
    if np.any(np.isfinite(depths)):
        j = int(np.nanargmin(np.abs(depths - float(target_depth))))
    else:
        j = 0
    return sensors[j]


def _scan_sensor_dataframe(sensor_obj):
    try:
        df = sensor_obj.data
    except Exception:
        df = sensor_obj.read_data()

    if not isinstance(df, pd.DataFrame):
        df = pd.DataFrame(df)

    if not isinstance(df.index, pd.DatetimeIndex):
        df.index = pd.to_datetime(df.index)
    return df


def _scan_pick_var_col(df, var_name):
    cols = [c for c in df.columns if var_name in str(c).lower()]
    if len(cols) > 0:
        return cols[0]
    num_cols = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]
    if len(num_cols) == 0:
        return None
    return num_cols[0]


def _scan_sensor_series_qc(sm_sensor_obj, required_sm_flag='G'):
    df_sm = _scan_sensor_dataframe(sm_sensor_obj)

    sm_col = _scan_pick_var_col(df_sm, 'soil_moisture')
    if sm_col is None:
        raise RuntimeError(f'No soil moisture numeric column found for sensor {getattr(sm_sensor_obj, "name", "")}')

    flag_cols = [c for c in df_sm.columns if str(c).lower() == 'soil_moisture_flag']
    if len(flag_cols) == 0:
        flag_cols = [c for c in df_sm.columns if 'soil_moisture_flag' in str(c).lower()]
    if len(flag_cols) == 0:
        raise RuntimeError(f'No soil_moisture_flag column found for sensor {getattr(sm_sensor_obj, "name", "")}')

    sm = pd.to_numeric(df_sm[sm_col], errors='coerce').copy()
    flag = df_sm[flag_cols[0]].astype(str).str.strip().str.upper()
    keep_mask = flag.eq(str(required_sm_flag).strip().upper())

    s = sm.where(keep_mask)
    s = s.groupby(s.index).mean().sort_index()
    return s


# load SCAN MAT data
scan_mat = sio.loadmat(SCAN_OL_MAT_PATH, squeeze_me=True, struct_as_record=False)
required = ['LDAS_sm_org', 'INSITU_sm', 'date_time_vec']
missing = [k for k in required if k not in scan_mat]
if missing:
    raise KeyError(f'SCAN MAT missing required keys: {missing}')

scan_ldas = np.array(scan_mat['LDAS_sm_org'], dtype=float)
scan_obs_mat = np.array(scan_mat['INSITU_sm'], dtype=float)
if scan_ldas.ndim == 4:
    scan_ldas = scan_ldas[:, :, :, 0]
if scan_ldas.ndim != 3 or scan_obs_mat.ndim != 3:
    raise ValueError(f'Unexpected SCAN array shapes: ldas={scan_ldas.shape}, obs={scan_obs_mat.shape}')

scan_ldas_surface = scan_ldas[:, 0, :].astype(float)
scan_obs_surface = scan_obs_mat[:, 0, :].astype(float)
scan_ldas_surface[scan_ldas_surface == 0] = np.nan
scan_obs_surface[scan_obs_surface < -9000] = np.nan
scan_time = _scan_plot_build_timestamps(scan_mat['date_time_vec'])

# choose random matched SCAN stations
cand = match_df_scan[np.isfinite(match_df_scan['match_distance_km'])].copy()
close_cand = cand[cand['match_distance_km'] <= SCAN_MAX_MATCH_KM].copy()
if len(close_cand) >= SCAN_N_RANDOM_SITES:
    cand = close_cand
if len(cand) == 0:
    raise RuntimeError('No SCAN matched stations with finite distance found.')

n_pick = min(SCAN_N_RANDOM_SITES, len(cand))
rng = np.random.default_rng(SCAN_RANDOM_SEED)
pick_idx = rng.choice(cand.index.to_numpy(), size=n_pick, replace=False)
sel_scan = cand.loc[pick_idx].copy().sort_values('ismn_station_name').reset_index(drop=True)

print('Selected random SCAN stations for comparison:')
display(sel_scan[[
    'ismn_station_name', 'ismn_lat', 'ismn_lon',
    'existing_site_index', 'existing_site_id', 'existing_lat', 'existing_lon', 'match_distance_km'
]])

scan_site_series = []
for _, row in sel_scan.iterrows():
    station_name = str(row['ismn_station_name'])
    site_idx = int(row['existing_site_index'])

    if site_idx < 0 or site_idx >= scan_ldas_surface.shape[1]:
        raise IndexError(f'existing_site_index {site_idx} out of bounds for SCAN MAT arrays')

    st = ismn_data['SCAN'][station_name]
    sm_sensor = _scan_pick_surface_sm_sensor(st, target_depth=SCAN_TARGET_SURFACE_DEPTH_M)
    s_ismn_native = _scan_sensor_series_qc(sm_sensor, required_sm_flag='G')
    s_ismn_daily = s_ismn_native.resample('1D').mean()
    s_ismn_daily.index = s_ismn_daily.index + pd.Timedelta(hours=12)

    s_ol_native = pd.Series(scan_ldas_surface[:, site_idx], index=scan_time).sort_index()
    s_obs_native = pd.Series(scan_obs_surface[:, site_idx], index=scan_time).sort_index()

    scan_site_series.append({
        'station_name': station_name,
        'site_idx': site_idx,
        'site_id': row['existing_site_id'],
        's_ismn_native': s_ismn_native,
        's_ismn_daily': s_ismn_daily,
        's_obs_native': s_obs_native,
        's_ol_native': s_ol_native,
    })

# Figure 1: full period
fig1, axs1 = plt.subplots(n_pick, 1, figsize=(14, 4 * n_pick), sharex=False)
if n_pick == 1:
    axs1 = [axs1]

for ax, dct in zip(axs1, scan_site_series):
    ax.plot(dct['s_ismn_native'].index, dct['s_ismn_native'].values, color='k', lw=0.8, alpha=0.8, label='ISMN Obs (native, G-flag QC)')
    ax.plot(dct['s_obs_native'].index, dct['s_obs_native'].values, color='#4C78A8', lw=1.0, label='MAT Obs (native)')
    ax.plot(dct['s_ol_native'].index, dct['s_ol_native'].values, color='#F28E2B', lw=1.0, label='OL model (native)')

    ax.set_title(f"SCAN {dct['station_name']}  <->  MAT site_id {dct['site_id']} (index {dct['site_idx']})")
    ax.set_ylabel('Surface SM')
    ax.grid(alpha=0.3)
    ax.legend(loc='upper right')

axs1[-1].set_xlabel('Time')
fig1.suptitle('Random matched SCAN sites: ISMN Obs vs MAT Obs vs OL (surface SM, native timesteps)', y=1.02)
plt.tight_layout()
plt.show()

# Figure 2: 2020-2021 zoom
fig2, axs2 = plt.subplots(n_pick, 1, figsize=(14, 4 * n_pick), sharex=False)
if n_pick == 1:
    axs2 = [axs2]

for ax, dct in zip(axs2, scan_site_series):
    s_ismn_daily_zoom = dct['s_ismn_daily'].loc[(dct['s_ismn_daily'].index >= SCAN_ZOOM_START) & (dct['s_ismn_daily'].index <= SCAN_ZOOM_END)]
    s_obs_zoom = dct['s_obs_native'].loc[(dct['s_obs_native'].index >= SCAN_ZOOM_START) & (dct['s_obs_native'].index <= SCAN_ZOOM_END)]

    ax.plot(s_ismn_daily_zoom.index, s_ismn_daily_zoom.values, lw=2.5, linestyle=":", label='ISMN Obs (daily mean, G-flag QC)')
    ax.plot(s_obs_zoom.index, s_obs_zoom.values, lw=1.0, label='MAT Obs (native)')

    ax.set_title(f"SCAN {dct['station_name']}  <->  MAT site_id {dct['site_id']} (index {dct['site_idx']}) | 2020-2021")
    ax.set_ylabel('Surface SM')
    ax.grid(alpha=0.3)
    ax.legend(loc='upper right')

axs2[-1].set_xlabel('Time')
fig2.suptitle('Random matched SCAN sites: 2020-2021 ISMN daily vs MAT Obs native', y=1.02)
plt.tight_layout()
plt.show()

```

```python
# Coverage comparison scatter: MAT vs ISMN daily days by site (USCRN + SCAN)
# Includes both ISMN no-QC and QC counts to diagnose filtering impact.
import re
import numpy as np
import pandas as pd
import scipy.io as sio
import matplotlib.pyplot as plt
from pathlib import Path

if 'ismn_data' not in globals():
    raise RuntimeError('Run the ISMN initialization cell first (ismn_data missing).')
if 'match_df' not in globals() or 'match_df_scan' not in globals():
    raise RuntimeError('Run both matching cells first (match_df and match_df_scan are required).')

USCRN_MAT_PATH = Path('/Users/amfox/Desktop/GEOSldas_diagnostics/test_data/M21C_land_sweeper_v2/LS_OLv8_M36_USCRN_SM_1d_c1234smv_25yr_raw_timeseries.mat')
SCAN_MAT_PATH = Path('/Users/amfox/Desktop/GEOSldas_diagnostics/test_data/M21C_land_sweeper_v2/LS_OLv8_M36_SCAN_SM_1d_c1234smv_25yr_raw_timeseries.mat')
TARGET_SURFACE_DEPTH_M = 0.05


def _cov_extract_depth_from_name(sensor_name):
    m = re.search(r'_(-?\d+\.\d+)_(-?\d+\.\d+)$', str(sensor_name))
    if not m:
        return np.nan
    d1 = float(m.group(1))
    d2 = float(m.group(2))
    return 0.5 * (d1 + d2)


def _cov_pick_sensor(station_obj, variable, target_depth=0.05):
    sensors = []
    if hasattr(station_obj, 'iter_sensors'):
        try:
            sensors = list(station_obj.iter_sensors(variable=variable))
        except TypeError:
            sensors = [s for s in station_obj.iter_sensors() if variable in getattr(s, 'name', '').lower()]

    if len(sensors) == 0:
        return None

    depths = np.array([_cov_extract_depth_from_name(getattr(s, 'name', '')) for s in sensors], dtype=float)
    if np.any(np.isfinite(depths)):
        j = int(np.nanargmin(np.abs(depths - float(target_depth))))
    else:
        j = 0
    return sensors[j]


def _cov_sensor_dataframe(sensor_obj):
    try:
        df = sensor_obj.data
    except Exception:
        df = sensor_obj.read_data()

    if not isinstance(df, pd.DataFrame):
        df = pd.DataFrame(df)

    if not isinstance(df.index, pd.DatetimeIndex):
        df.index = pd.to_datetime(df.index)

    return df


def _cov_pick_var_col(df, var_name):
    cols = [c for c in df.columns if var_name in str(c).lower()]
    if len(cols) > 0:
        return cols[0]
    num_cols = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]
    if len(num_cols) == 0:
        return None
    return num_cols[0]


def _cov_ismn_daily_no_qc(station_obj, target_depth=0.05):
    sm_sensor = _cov_pick_sensor(station_obj, 'soil_moisture', target_depth=target_depth)
    if sm_sensor is None:
        raise RuntimeError('No soil_moisture sensor found')

    df_sm = _cov_sensor_dataframe(sm_sensor)
    sm_col = _cov_pick_var_col(df_sm, 'soil_moisture')
    if sm_col is None:
        raise RuntimeError('No soil_moisture numeric column found')

    s_native = pd.to_numeric(df_sm[sm_col], errors='coerce').groupby(df_sm.index).mean().sort_index()
    s_daily = s_native.resample('1D').mean()
    return s_daily


def _cov_ismn_daily_qc(station_obj, target_depth=0.05):
    sm_sensor = _cov_pick_sensor(station_obj, 'soil_moisture', target_depth=target_depth)
    if sm_sensor is None:
        raise RuntimeError('No soil_moisture sensor found')

    df_sm = _cov_sensor_dataframe(sm_sensor)
    sm_col = _cov_pick_var_col(df_sm, 'soil_moisture')
    if sm_col is None:
        raise RuntimeError('No soil_moisture numeric column found')

    flag_cols = [c for c in df_sm.columns if str(c).lower() == 'soil_moisture_flag']
    if len(flag_cols) == 0:
        flag_cols = [c for c in df_sm.columns if 'soil_moisture_flag' in str(c).lower()]
    if len(flag_cols) == 0:
        raise RuntimeError('No soil_moisture_flag column found')

    sm = pd.to_numeric(df_sm[sm_col], errors='coerce').copy()
    keep_mask = df_sm[flag_cols[0]].astype(str).str.strip().str.upper().eq('G')

    s_native_qc = sm.where(keep_mask).groupby(sm.index).mean().sort_index()
    s_daily_qc = s_native_qc.resample('1D').mean()
    return s_daily_qc


def _cov_load_mat_obs_surface(mat_path):
    if not mat_path.exists():
        raise FileNotFoundError(f'MAT file not found: {mat_path}')

    mat = sio.loadmat(mat_path, squeeze_me=True, struct_as_record=False)
    if 'INSITU_sm' not in mat:
        raise KeyError(f'INSITU_sm missing in MAT file: {mat_path}')

    obs = np.array(mat['INSITU_sm'], dtype=float)
    if obs.ndim == 4:
        obs = obs[:, :, :, 0]
    if obs.ndim != 3:
        raise ValueError(f'Unexpected INSITU_sm shape for {mat_path.name}: {obs.shape}')

    obs_surface = obs[:, 0, :].astype(float)
    obs_surface[obs_surface < -9000] = np.nan
    return obs_surface


def _coverage_df(network, match_table, mat_path):
    obs_surface = _cov_load_mat_obs_surface(mat_path)

    rows = []
    for _, row in match_table.iterrows():
        site_idx = int(row['existing_site_index'])
        station_name = str(row['ismn_station_name'])

        if site_idx < 0 or site_idx >= obs_surface.shape[1]:
            continue

        mat_days = int(np.isfinite(obs_surface[:, site_idx]).sum())

        ismn_days_no_qc = np.nan
        ismn_days_qc = np.nan
        read_ok_no_qc = False
        read_ok_qc = False

        try:
            st = ismn_data[network][station_name]

            s_ismn_daily_no_qc = _cov_ismn_daily_no_qc(
                st,
                target_depth=TARGET_SURFACE_DEPTH_M,
            )
            ismn_days_no_qc = int(np.isfinite(s_ismn_daily_no_qc.to_numpy(dtype=float)).sum())
            read_ok_no_qc = True

            s_ismn_daily_qc = _cov_ismn_daily_qc(
                st,
                target_depth=TARGET_SURFACE_DEPTH_M,
            )
            ismn_days_qc = int(np.isfinite(s_ismn_daily_qc.to_numpy(dtype=float)).sum())
            read_ok_qc = True

        except Exception:
            pass

        days_removed_by_qc = np.nan
        qc_retention_pct = np.nan
        if np.isfinite(ismn_days_no_qc) and np.isfinite(ismn_days_qc):
            days_removed_by_qc = float(ismn_days_no_qc - ismn_days_qc)
            if ismn_days_no_qc > 0:
                qc_retention_pct = 100.0 * float(ismn_days_qc) / float(ismn_days_no_qc)

        rows.append({
            'network': network,
            'ismn_station_name': station_name,
            'existing_site_id': row.get('existing_site_id', np.nan),
            'existing_site_index': site_idx,
            'match_distance_km': float(row.get('match_distance_km', np.nan)),
            'mat_days': mat_days,
            'ismn_days_no_qc': ismn_days_no_qc,
            'ismn_days_qc': ismn_days_qc,
            'days_removed_by_qc': days_removed_by_qc,
            'qc_retention_pct': qc_retention_pct,
            'ismn_read_ok_no_qc': read_ok_no_qc,
            'ismn_read_ok_qc': read_ok_qc,
        })

    out = pd.DataFrame(rows)
    return out.sort_values('ismn_station_name').reset_index(drop=True)


uscrn_cov = _coverage_df('USCRN', match_df, USCRN_MAT_PATH)
scan_cov = _coverage_df('SCAN', match_df_scan, SCAN_MAT_PATH)

fig, axs = plt.subplots(1, 2, figsize=(15, 6), sharex=False, sharey=False)

for ax, df_cov, net_name in zip(axs, [uscrn_cov, scan_cov], ['USCRN', 'SCAN']):
    valid_no_qc = df_cov[np.isfinite(df_cov['mat_days']) & np.isfinite(df_cov['ismn_days_no_qc'])].copy()
    valid_qc = df_cov[np.isfinite(df_cov['mat_days']) & np.isfinite(df_cov['ismn_days_qc'])].copy()

    ax.scatter(
        valid_no_qc['mat_days'].to_numpy(dtype=float),
        valid_no_qc['ismn_days_no_qc'].to_numpy(dtype=float),
        s=34,
        alpha=0.65,
        edgecolor='none',
        color='0.65',
        label='ISMN no-QC',
    )

    ax.scatter(
        valid_qc['mat_days'].to_numpy(dtype=float),
        valid_qc['ismn_days_qc'].to_numpy(dtype=float),
        s=34,
        alpha=0.75,
        edgecolor='none',
        color='#4C78A8',
        label='ISMN QC',
    )

    vmax = 1.0
    if len(valid_no_qc) > 0:
        vmax = max(vmax, float(np.nanmax(valid_no_qc[['mat_days', 'ismn_days_no_qc']].to_numpy(dtype=float))))
    if len(valid_qc) > 0:
        vmax = max(vmax, float(np.nanmax(valid_qc[['mat_days', 'ismn_days_qc']].to_numpy(dtype=float))))

    ax.plot([0, vmax], [0, vmax], '--', color='0.4', lw=1)
    ax.set_xlim(left=0)
    ax.set_ylim(bottom=0)

    med_ret = np.nanmedian(df_cov['qc_retention_pct'].to_numpy(dtype=float)) if len(df_cov) > 0 else np.nan
    ax.set_title(f'{net_name}: n={len(df_cov)}, median QC retention={med_ret:.1f}%')
    ax.set_xlabel('MAT Obs days')
    ax.set_ylabel('ISMN Obs days')
    ax.grid(alpha=0.3)
    ax.legend(loc='lower right')

fig.suptitle('Data coverage by site: MAT vs ISMN (no-QC and QC)', y=1.02)
plt.tight_layout()
plt.show()

for net_name, df_cov in [('USCRN', uscrn_cov), ('SCAN', scan_cov)]:
    print(f'{net_name} summary:')
    if len(df_cov) == 0:
        print('  no matched rows')
        continue

    n_no_qc = int(np.isfinite(df_cov['ismn_days_no_qc']).sum())
    n_qc = int(np.isfinite(df_cov['ismn_days_qc']).sum())
    med_no_qc = float(np.nanmedian(df_cov['ismn_days_no_qc'])) if n_no_qc > 0 else np.nan
    med_qc = float(np.nanmedian(df_cov['ismn_days_qc'])) if n_qc > 0 else np.nan
    med_removed = float(np.nanmedian(df_cov['days_removed_by_qc'])) if n_qc > 0 else np.nan
    med_ret = float(np.nanmedian(df_cov['qc_retention_pct'])) if n_qc > 0 else np.nan

    print(f'  sites with ISMN no-QC counts: {n_no_qc}/{len(df_cov)}')
    print(f'  sites with ISMN QC counts: {n_qc}/{len(df_cov)}')
    print(f'  median ISMN days (no-QC): {med_no_qc:.1f}')
    print(f'  median ISMN days (QC): {med_qc:.1f}')
    print(f'  median days removed by QC: {med_removed:.1f}')
    print(f'  median QC retention (%): {med_ret:.1f}')

print('USCRN coverage table (head):')
display(uscrn_cov.head(20))
print('SCAN coverage table (head):')
display(scan_cov.head(20))
```

```python
# QC-flag distribution for soil moisture files: USCRN and SCAN
from pathlib import Path
from collections import Counter
import numpy as np
import pandas as pd

ISMN_ROOT = Path('/Users/amfox/Desktop/ISMN_data')
FLAGS_TXT = ISMN_ROOT / 'ISMN_qualityflags_description.txt'
NETWORKS = ['USCRN', 'SCAN']
TOP_COMBINED = 20

if not FLAGS_TXT.exists():
    raise FileNotFoundError(f'Flag description file not found: {FLAGS_TXT}')


def _load_sm_flag_meanings(flags_txt_path):
    meanings = {}
    with flags_txt_path.open('r', encoding='utf-8', errors='ignore') as f:
        for raw in f:
            if ';' not in raw:
                continue
            parts = [p.strip() for p in raw.strip().split(';')]
            if len(parts) < 3:
                continue

            variable = parts[0].lower()
            if variable != 'soil moisture':
                continue

            # Typical format: variable;flag category;flag;condition
            if len(parts) >= 4:
                code = parts[2].strip()
                meaning = parts[3].strip()
            else:
                # Fallback if category column is absent
                code = parts[1].strip()
                meaning = parts[2].strip()

            if code == '' or code.lower() == 'flag':
                continue
            meanings[code] = meaning
    return meanings


def _soil_moisture_flag_distribution(ismn_root, network):
    network_dir = ismn_root / network
    sm_files = sorted(network_dir.rglob('*_sm_*.stm'))
    if len(sm_files) == 0:
        raise RuntimeError(f'No soil moisture files found under: {network_dir}')

    individual = Counter()
    combined = Counter()
    total_records = 0

    for fp in sm_files:
        with fp.open('r', encoding='utf-8', errors='ignore') as f:
            next(f, None)  # metadata/header line
            for line in f:
                parts = line.split()
                if len(parts) < 4:
                    continue

                total_records += 1
                flag_str = parts[3].strip()
                if flag_str == '':
                    continue

                combined[flag_str] += 1
                for code in flag_str.split(','):
                    code = code.strip()
                    if code:
                        individual[code] += 1

    return sm_files, total_records, individual, combined


flag_meanings = _load_sm_flag_meanings(FLAGS_TXT)

qc_flag_results = {}

for net in NETWORKS:
    files, total_records, indiv_counter, combo_counter = _soil_moisture_flag_distribution(ISMN_ROOT, net)

    indiv_df = pd.DataFrame([
        {
            'flag': code,
            'count': int(n),
            'pct_of_records': 100.0 * float(n) / float(total_records) if total_records > 0 else np.nan,
            'meaning': flag_meanings.get(code, '(not found in ISMN_qualityflags_description.txt)'),
        }
        for code, n in indiv_counter.items()
    ]).sort_values('count', ascending=False).reset_index(drop=True)

    combo_df = pd.DataFrame([
        {
            'flag_string': combo,
            'count': int(n),
            'pct_of_records': 100.0 * float(n) / float(total_records) if total_records > 0 else np.nan,
        }
        for combo, n in combo_counter.items()
    ]).sort_values('count', ascending=False).reset_index(drop=True)

    qc_flag_results[net] = {
        'n_files': len(files),
        'n_records': int(total_records),
        'individual_df': indiv_df,
        'combined_df': combo_df,
    }

    print(f'\n{net} soil-moisture QC flags')
    print(f'  files: {len(files)}')
    print(f'  records: {int(total_records)}')
    print('  Note: individual-flag percentages can sum >100% because combined flags are split by code.')

    print('\nIndividual flags (all):')
    display(indiv_df)

    print(f'Combined flag strings (top {TOP_COMBINED}):')
    display(combo_df.head(TOP_COMBINED))
```

```python
# SCAN QC diagnostics: isolate impact of G-flag filtering only
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

if 'ismn_data' not in globals() or 'match_df_scan' not in globals():
    raise RuntimeError('Run ISMN init and SCAN matching cells first (ismn_data + match_df_scan required).')

DIAG_DEPTH_M = 0.05


def _diag_extract_depth(sensor_name):
    m = re.search(r'_(-?\d+\.\d+)_(-?\d+\.\d+)$', str(sensor_name))
    if not m:
        return np.nan
    return 0.5 * (float(m.group(1)) + float(m.group(2)))


def _diag_pick_sensor(station_obj, variable, target_depth=0.05):
    sensors = []
    if hasattr(station_obj, 'iter_sensors'):
        try:
            sensors = list(station_obj.iter_sensors(variable=variable))
        except TypeError:
            sensors = [s for s in station_obj.iter_sensors() if variable in getattr(s, 'name', '').lower()]

    if len(sensors) == 0:
        return None

    depths = np.array([_diag_extract_depth(getattr(s, 'name', '')) for s in sensors], dtype=float)
    if np.any(np.isfinite(depths)):
        j = int(np.nanargmin(np.abs(depths - float(target_depth))))
    else:
        j = 0
    return sensors[j]


def _diag_df(sensor_obj):
    try:
        df = sensor_obj.data
    except Exception:
        df = sensor_obj.read_data()
    if not isinstance(df, pd.DataFrame):
        df = pd.DataFrame(df)
    if not isinstance(df.index, pd.DatetimeIndex):
        df.index = pd.to_datetime(df.index)
    return df


def _diag_pick_col(df, var_name):
    cols = [c for c in df.columns if var_name in str(c).lower()]
    if len(cols) > 0:
        return cols[0]
    num_cols = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c])]
    return num_cols[0] if len(num_cols) > 0 else None


def _daily_count(s):
    if s is None or len(s) == 0:
        return 0
    return int(np.isfinite(s.resample('1D').mean().to_numpy(dtype=float)).sum())


diag_rows = []
for _, row in match_df_scan.iterrows():
    station_name = str(row['ismn_station_name'])

    try:
        st = ismn_data['SCAN'][station_name]

        sm_sensor = _diag_pick_sensor(st, 'soil_moisture', target_depth=DIAG_DEPTH_M)
        if sm_sensor is None:
            continue

        df_sm = _diag_df(sm_sensor)
        sm_col = _diag_pick_col(df_sm, 'soil_moisture')
        if sm_col is None:
            continue

        flag_cols = [c for c in df_sm.columns if str(c).lower() == 'soil_moisture_flag']
        if len(flag_cols) == 0:
            flag_cols = [c for c in df_sm.columns if 'soil_moisture_flag' in str(c).lower()]
        if len(flag_cols) == 0:
            continue

        sm_rec = pd.to_numeric(df_sm[sm_col], errors='coerce')
        flag_rec = df_sm[flag_cols[0]].astype(str).str.strip().str.upper()
        g_rec = flag_rec.eq('G')

        s_no_qc = sm_rec.groupby(sm_rec.index).mean().sort_index()
        s_g_only = sm_rec.where(g_rec).groupby(sm_rec.index).mean().sort_index()

        d_no_qc = _daily_count(s_no_qc)
        d_g = _daily_count(s_g_only)

        diag_rows.append({
            'ismn_station_name': station_name,
            'match_distance_km': float(row.get('match_distance_km', np.nan)),
            'n_records': int(len(sm_rec)),
            'n_sm_non_nan': int(sm_rec.notna().sum()),
            'n_g_records': int((sm_rec.notna() & g_rec).sum()),
            'days_no_qc': d_no_qc,
            'days_g_only': d_g,
            'days_lost_noqc_to_g': int(d_no_qc - d_g),
        })

    except Exception:
        continue

scan_qc_diag_df = pd.DataFrame(diag_rows).sort_values('ismn_station_name').reset_index(drop=True)

if len(scan_qc_diag_df) == 0:
    raise RuntimeError('No SCAN QC diagnostic rows produced.')

print('SCAN QC diagnostics summary (G-flag only)')
print(f'  matched stations analyzed: {len(scan_qc_diag_df)}')
print(f'  median days (no QC): {np.nanmedian(scan_qc_diag_df["days_no_qc"]):.1f}')
print(f'  median days (G only): {np.nanmedian(scan_qc_diag_df["days_g_only"]):.1f}')

print('')
print('Sites with largest day loss from no-QC to G-only:')
display(
    scan_qc_diag_df[
        ['ismn_station_name', 'match_distance_km', 'days_no_qc', 'days_g_only', 'days_lost_noqc_to_g']
    ]
    .sort_values('days_lost_noqc_to_g', ascending=False)
    .head(25)
)

fig, ax = plt.subplots(1, 1, figsize=(7, 5))
x = scan_qc_diag_df['days_no_qc'].to_numpy(dtype=float)
y = scan_qc_diag_df['days_g_only'].to_numpy(dtype=float)
ax.scatter(x, y, s=35, alpha=0.75, color='#4C78A8', edgecolor='none')
if np.isfinite(x).any() and np.isfinite(y).any():
    vmax = float(np.nanmax(np.r_[x, y, 1.0]))
    ax.plot([0, vmax], [0, vmax], '--', color='0.4', lw=1)
ax.set_xlabel('ISMN days (no QC)')
ax.set_ylabel('ISMN days (G-only QC)')
ax.set_title('G-flag QC impact')
ax.grid(alpha=0.3)
fig.suptitle('SCAN: diagnosing G-flag filtering only', y=1.02)
plt.tight_layout()
plt.show()
```
